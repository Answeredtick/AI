### My Question
Among all the adults surveyed, how many smoke compared to how many drink alcohol

### My Study
It is a census study because we are comparing the entire population and how many answered yes to drinking alcohol to how many answered yes to smoking cigarettes

### My Analysis

```python
import pandas as pd
```

```python
import seaborn
```

```python
df = pd.read_csv("Steak.csv")
```

```python
df.head(15)
```

```python
S = (df[df["Do you ever smoke cigarettes?"] == "Yes"])
```

```python
D = (df[df["Do you ever drink alcohol?"] == "Yes"])
```

```python
count_S= len(S)
print("Smoke: ", count_S)

count_D= len(D)
print("Drink: ", count_D)
```

```python
df["Do you ever smoke cigarettes?"].value_counts()
```

```python
df["Do you ever drink alcohol?"].value_counts()
```

```python
seaborn.countplot(data=df, x="Do you ever smoke cigarettes?")
```

```python
seaborn.countplot(data=df, x="Do you ever drink alcohol?")
```

```python
counts = pd.DataFrame({
    "Response": ["Yes", "No"],
    "Smoke": [
        df["Do you ever smoke cigarettes?"].value_counts()["Yes"],
        df["Do you ever smoke cigarettes?"].value_counts()["No"]
    ],
    "Drink alcohol": [
        df["Do you ever drink alcohol?"].value_counts()["Yes"],
        df["Do you ever drink alcohol?"].value_counts()["No"]
    ]
})

counts = counts.melt(id_vars="Response",
                     var_name="Behavior",
                     value_name="Count")

seaborn.barplot(data=counts, x="Response", y="Count", hue="Behavior")
```
