# Question

***How does the median income of families with kids going to boulder or Mines compare to families with students at other colleges***

This is a census study because i will look at all colledges and compare the median family income to see how the median family income of boulder students and mines students compare to other schools

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("cleaned_college.csv")
```

```python
df.head(15)
```

```python
boulder=df[df["school_name"]=="University of Colorado Boulder"]
```

```python
mines=df[df["school_name"]=="Colorado School of Mines"]
```

```python
print(boulder["median_family_income"])
print(mines["median_family_income"])
```

```python
sns.histplot(data=df, x="median_family_income")
plt.axvline(x=95766.0, linewidth=1, linestyle="--", color='red', label="mines")
plt.axvline(x=70221.0, linewidth=1, linestyle="--", color='blue', label="boulder")
plt.legend()
```

```python
g=df["median_family_income"].mean()
g2=df["median_family_income"].median()
g3=df["median_family_income"].std()
```

```python
print(g)
print(g2)
print(g3)
```

```python
(boulder["median_family_income"].item()-g)/g3
```

```python
(mines["median_family_income"].item()-g)/g3
```

# conclusion:

The University of Colorado Boulder has a z score of 1.99, sitting around 2 standard deviations above the average median family income, while the School of Mines has a z score of 3.26, sitting as an outlier



As shown in the graph, Mines is closer to the end of the tail, while Boulder is closer to the middle
