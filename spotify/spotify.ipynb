# Question:
***what is the most common BPM of popular songs up to 2023 songs?***

This is an observational study because we are looking at the most popular songs of 2023 from Spotify, which is only a sample of the popular songs of 2023, without changing anything in the data

```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv("spotify-2023.csv", encoding="latin-1")
```

```python
df.head(15)
```

```python
df.info()
```

```python
df["bpm"]
```

```python
mini=df["bpm"].min()
```

```python
mean=df["bpm"].mean()
```

```python
med=df["bpm"].median()
```

```python
quant=df["bpm"].quantile()
```

```python
maxi=df["bpm"].max()
```

```python
print("5-Number Summary of BPM for songs released up to 2023")
print("min: ", mini)
print("mean: ", mean)
print("median: ", med)
print("quantile: ", quant)
print("max: ", maxi)
```

```python
g=sns.histplot(data=df, x="bpm")
```

#  Conclusion: 

The average BPM of popular songs up to 2023 is 122 beats per minute



This graph uses the bpm of the songs from the provided data to show their approximate bpm, with the most commonly used bpm sitting between 120 and 130 bpm
