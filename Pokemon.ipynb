# Question:

***how does pecharunt's defense compare to other mythical pokemon's***

# my study

I will conduct a census study because i am only looking at the mythical Pokémon out of the full Pokedex

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("pokemon.csv")
```

```python
df.head(15)
```

```python
df.info()
```

```python
is_m=df[df["is_mythical"]==1]
```

```python
plt.axvline(x=96.52, linewidth=1, linestyle="--", color='red', label="mean")
plt.axvline(x=100, linewidth=1, linestyle="--", color='blue', label="median")
sns.histplot(data=is_m, x="defense")
plt.legend()
```

```python
std_defense=is_m["defense"].std()
mean_defense=is_m["defense"].mean()
median_defense=is_m["defense"].median()
```

```python
one_std_above=std_defense+mean_defense
```

```python
print(one_std_above)
```

```python
plt.axvline(x=160, linewidth=1, linestyle="--", color='violet', label="Pecharunt")
plt.axvline(x=96.52, linewidth=1, linestyle="--", color='red', label="mean")
plt.axvline(x=100, linewidth=1, linestyle="--", color='blue', label="median")
sns.boxplot(data=is_m, x="defense")
plt.legend()
```

```python
print(mean_defense)
print(median_defense)
```
