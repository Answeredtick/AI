### My Question

***How many times will 1 be rolled on a 20-sided die in 1000 rolls, and what is the probability of it?***

### My Study

We will be simulating a dice roll our trial will look like us rolling a die which will produce a outcome of a number in the range of 1 to 20 we will be looking at how many times that outcome is 1

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
### My Analysis
Rolls = np.random.randint(1,21,size=1000)
```

```python
df=pd.DataFrame({'Roll':Rolls})
```

```python
df.head(1000)
```

```python
frequency=df['Roll'].value_counts().sort_index()
```

```python
print(frequency)
```

```python
ones = (df['Roll']==1).sum()
```

```python
prob_one=(df['Roll']==1).mean()
```

```python
print("Number of ones:", ones)
print("Empirical probability:", prob_one)
```

```python
relative = df['Roll'].value_counts(normalize=True).sort_index()
print(relative)
# 1/20 = .05, which is our expected probability
```

```python
theory_prob = 1 / 20
```

```python
print(theory_prob)
```

```python
sns.countplot(data=df, x='Roll')
plt.xlabel('Die Face')
plt.ylabel('Frequency')
plt.show()
```

```python
summary = pd.DataFrame({'Observed Frequency': frequency,'Observed Probability': relative,'Expected Probability': 0.05})
```

```python
print(summary)
```

### My Conclusion
We rolled 1 59 times out of our thousand, giving us an empirical probability of 5.9% while our theoretical was .5%



The graph shows how many times each face was rolled, and 1 is shown to have been rolled slightly less than sixty times
