# Question:
***Is someone's favorite season associated with their favorite kind of music***

I will conduct a Observational study because i am only looking at teens in colorado to find my answer and only have access to a sample of data collected from the ASA. which is not representive of every student in the United States

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("CensusAtSchool_Sample.csv")
```

```python
df.head(20)
```

```python
plt.figure(figsize=(30, 6))
sns.countplot(df,x="Favorite_Music", hue="Favorite_Season")
```

```python
df["Favorite_Music"].value_counts()
```

```python
df["Favorite_Season"].value_counts()
```

```python
print(df["Favorite_Music"])
```

# Conclusion:

there appears to be a small association between favorite season and favorite kind of music because the conditional relative frequencies are not the same across all seasons for example rap/hip hop is more popular with 33.8% and 34.2% of people who like it also liking spring and summer while pop is more popular among people who like fall with 24.6% of people, showing that favorite kind music and favorite season may not be completely independent of each other.



The graph shows approximately how many people like the four seasons and their favorite season, letting us see which seasons are favorites across different music genres.
